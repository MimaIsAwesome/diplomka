```python
import cv2
import os
import time
import uuid
```

```python
IMAGES_PATH = 'Tensorflow/workspace/images/collectedimages'
```

```python
labels_alpha = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k',
          'l', 'm', 'n', 'o', 'p', 'q', 'r', 's', 't', 'u', 'v', 'w', 'x', 'y', 'z'] # ABECEDA done
labels_num = ['1', '2', '3', '4', '5', '6', '7', '8', '9'] #NUMBERS done
labels_verbs = ['chciet', 'davat', 'drzat', 'jest', 'lezat', 'prosit', 'ucit', 'usmievat'] #VERBS done
labels_nouns = ['auto', 'brat', 'budova', 'chlapec', 'dievca', 'domov', 'dvere', 'frajer',
          'hasic', 'kamera', 'krava', 'lekar', 'mama', 'pivo', 'policia', 'sestra', 'zena'] #NOUNS
labels_others = ['ako', 'alebo', 'ano', 'co', 'kedy', 'kto','preco'] #OTHERS
labels_pronouns = ['ja', 'ty', 'on'] #PRONOUNS

number_imgs = 20
```

```python
for label in labels_pronouns:
    !mkdir {'Tensorflow\workspace\images\collectedimages\\'+label}
    cap = cv2.VideoCapture(0)
    print('Collectiong images for {}'.format(label))
    time.sleep(10)
    for imagnum in range(number_imgs):
        ret, frame = cap.read()
        #print('selfieIdzes')
        imagename = os.path.join(IMAGES_PATH, label, label+'.'+'{}.jpg'.format(str(uuid.uuid1())))
        cv2.imwrite(imagename, frame)
        cv2.imshow('frame', frame)
        time.sleep(2)
        cv2.destroyWindow("frame")

        if cv2.waitKey(0) & 0xFF == ord('q'):
            break
    cap.release()
```
